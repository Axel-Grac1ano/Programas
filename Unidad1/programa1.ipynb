```python
tabla = int(input("dame la tabla que quieres"))
inicio = int(input("dame el inicio de la tabla"))
final = int(input("dame el final de la tabla"))
incrementos = int(input("dame el numero de los incrementos"))
while inicio < final:
  print("el limite final de la tabla debe ser mayor que el inicial")
  final = int(input("dame el limite fial"))
for i in range(inicio,final,incrementos):
  print(tabla, "*" , i, "=", tabla*i)
```

# **Listas en python**

```python
alumnos=["ana","luis","carlos"]
print(alumnos)
print(alumnos[0])
print(alumnos[2])
```

```python
print(alumnos)
alumnos[1]="juan"
print(alumnos)
```

```python
print(alumnos)
print(alumnos[-1])
print(alumnos[-2])
print(alumnos[-3])
```

```python
print(alumnos)
alumnos.append("sandra")
print(alumnos)
```

```python
print('Lista original:', alumnos)
alumnos.sort()
print('Lista ordenada alfabéticamente:', alumnos)
```
