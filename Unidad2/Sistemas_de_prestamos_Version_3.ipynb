```python
# ======================================
# SISTEMA DE PRESTAMOS
# Version 4 - Alumnos con nombre y carrera, lista de necesidades
# =====================================

from datetime import datetime

# Cada alumno: [num_control, nombre, carrera]
alumnos = [
    ["26040200", "Superman", "Ingeniería en Inteligencia Artificial"],
    ["26040201", "Bat Man", "Ingeniería en Inteligencia Artificial"],
    ["26040202", "Wonder Woman", "Ingeniería en Sistemas Computacionales"],
    ["26040203", "Flash", "Ingeniería en Sistemas Computacionales"],
    ["26040204", "Aquaman", "Ingeniería en Inteligencia Artificial"]
]

# Cada computadora: [ID, aula, software instalado, estado]
computadoras = [
    ["PC-LC1-01", "LC1", "Python, VS Code", "Disponible"],
    ["PC-LC1-02", "LC1", "Python, Java, VS Code", "Prestada"],
    ["PC-LC1-03", "LC1", "AutoCAD, MATLAB", "Reservada"],
    ["PC-LC2-01", "LC2", "Python, VS Code", "Disponible"],
    ["PC-LC2-02", "LC2", "Python, Java, VS Code", "Prestada"],
    ["PC-LC2-03", "LC2", "AutoCAD, MATLAB", "Reservada"],
    ["PC-LC3-01", "LC3", "Python, VS Code", "Disponible"],
    ["PC-LC3-02", "LC3", "Python, Java, VS Code", "Prestada"],
    ["PC-LC3-03", "LC3", "AutoCAD, MATLAB, Python, VS Code", "Reservada"],
    ["PC-LC3-04", "LC3", "Python, VS Code", "Mantenimiento"]
]

# Historial de prestamos: [num_control, programa, aula, ID]
prestamos = []

# Lista de necesidades de software: [programa, fecha y hora]
necesidades = []


def listar_computadoras():
  print("\n--- COMPUTADORAS ---")
  for i in range(len(computadoras)):
    print(i+1, " Aula:", computadoras[i][1], " | ID:", computadoras[i][0],
          "| Software:", computadoras[i][2], " |Estado:", computadoras[i][3])


def buscar_sw():
  programa = input("Ingrese el software que desea buscar: ")
  print("\n--- RESULTADOS ---")
  encontrado = False
  for i in range(len(computadoras)):
    if programa.lower() in computadoras[i][2].lower():
      print(i+1, " Aula:", computadoras[i][1], " | ID:", computadoras[i][0],
            "| Software:", computadoras[i][2], " |Estado:", computadoras[i][3])
      encontrado = True
  if encontrado == False:
    print("No se encontraron computadoras con ese software")


def solicitar_prestamo():
  print("\n--- SOLICITAR PRÉSTAMO ---")
  num_control = input("Número de control del alumno: ")

  # Ahora alumnos es una matriz: se compara con la primera columna alumnos[i][0]
  alumno_registrado = False
  for i in range(len(alumnos)):
    if alumnos[i][0] == num_control:
      alumno_registrado = True
      break

  if alumno_registrado:
    programa = input("Que programa necesita: ")
    software_encontrado = False
    asignada = False
    for i in range(len(computadoras)):
      if programa.lower() in computadoras[i][2].lower():
        software_encontrado = True
        if computadoras[i][3] == "Disponible":
          computadoras[i][3] = "Prestada"
          nuevo_prestamo = [num_control, programa, computadoras[i][1], computadoras[i][0]]
          prestamos.append(nuevo_prestamo)
          print("Se asigno la computadora", computadoras[i][0],
                "que esta en el aula", computadoras[i][1])
          asignada = True
          break

    if asignada == False:
      if software_encontrado:
        print("Si se encontro el software pero no hay computadoras disponibles")
      else:
        # El programa no existe: se pregunta antes de agregarlo,
        # porque pudo ser un error de escritura
        print("El programa", programa, "no existe en ninguna computadora.")
        respuesta = input("¿Desea agregarlo a la lista de necesidades? (s/n): ")
        if respuesta.lower() == "s":
          fecha = datetime.now()
          nueva_necesidad = [programa, fecha]
          necesidades.append(nueva_necesidad)
          print("Programa agregado a la lista de necesidades.")
        else:
          print("No se agrego a la lista de necesidades.")
  else:
    print("El alumno no esta registrado")


def devolver_computadora():
  print("\n--- DEVOLVER COMPUTADORA ---")
  id_computadora = input("Ingrese el ID de la computadora: ")
  existe = False
  for i in range(len(computadoras)):
    if computadoras[i][0] == id_computadora:
      existe = True
      if computadoras[i][3] == "Prestada" or computadoras[i][3] == "Reservada":
        computadoras[i][3] = "Disponible"
        print("La computadora", id_computadora, "ya esta disponible")
      else:
        print("La computadora", id_computadora, "no esta prestada/reservada")
      break
  if existe == False:
    print("No existe la computadora", id_computadora)


def agregar_computadora():
  print("\n--- AGREGAR COMPUTADORA ---")
  id_computadora = input("Ingrese el ID de la computadora: ")
  aula = input("Aula: ")
  programas = input("Software instalado: ")
  nueva_compu = [id_computadora, aula, programas, "Disponible"]
  computadoras.append(nueva_compu)
  print("Computadora agregada correctamente.")


def historial_prestamos():
  print("\n--- HISTORIAL DE PRESTAMOS DE COMPUTADORAS ---")
  if len(prestamos) == 0:
    print("Aun no hay prestamos registrados")
  for i in range(len(prestamos)):
    print(i+1, "| Alumno:", prestamos[i][0],
          "| Software:", prestamos[i][1],
          "| Aula:", prestamos[i][2],
          "| ID:", prestamos[i][3])


def necesidades_software():
  print("\n--- HISTORIAL DE NECESIDADES ---")
  if len(necesidades) == 0:
    print("Aun no hay necesidades registradas")
  for i in range(len(necesidades)):
    print(i+1, "| Programa:", necesidades[i][0],
          "| Fecha:", necesidades[i][1], "|")

def estadisticas():
  #conteo por carrera
  conteo_carreras = []
  for i in range(len(prestamos)):
    num_control = prestamos[i][0]
    carrera = ""
    for j in range(len(alumnos)):
      if alumnos[j][0] == num_control:
        carrera = alumnos[j][2]
  # contar la carrera
  encintrado = False
  for k in range(len(conteo_carreras)):
    if conteo_carreras[k][0] == carrera:
      conteo_carreras[k][1] += 1
      encontrado = True

  if not encontrado:
    conteo_carreras.append([carrera, 1])

# MENU PRINCIPAL
# El ciclo se repite hasta que el usuario teclee 9 (salir) o un numero mayor
opcion = 0
while opcion <= 8:
  print("\n================================")
  print("SISTEMA DE PRESTAMOS")
  print("================================")
  print("1. Consultar computadoras")
  print("2. Buscar por software")
  print("3. Solicitar prestamo")
  print("4. Devolver computadora")
  print("5. Crear reservacion")
  print("6. Agregar computadora")
  print("7. Historial de prestamos")
  print("8. Necesidades de Software")
  print("9. Estadisticas")
  print("10. Salir")
  print("================================")
  opcion = int(input("Seleccione una opcion: "))

  if opcion == 1:
    listar_computadoras()

  elif opcion == 2:
    buscar_sw()

  elif opcion == 3:
    solicitar_prestamo()

  elif opcion == 4:
    devolver_computadora()

  elif opcion == 5:
    print("pendiente.")

  elif opcion == 6:
    agregar_computadora()

  elif opcion == 7:
    historial_prestamos()

  elif opcion == 8:
    necesidades_software()

  elif opcion == 9:
    estadisticas()

  elif opcion >= 10:
    print("\nGracias por utilizar el sistema.")

  else:
    print("\nOpción no válida.")
```
